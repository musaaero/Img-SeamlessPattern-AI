```python
%%capture
import os
import pandas as pd
import requests
from PIL import Image
from io import BytesIO
from concurrent.futures import ThreadPoolExecutor, as_completed

# File paths
csv_file = "/kaggle/input/seamless-pattern/seamless_pattern.csv"
save_folder = "/kaggle/working/peft"

# Create the folder if it doesn't exist
os.makedirs(save_folder, exist_ok=True)

# Read the CSV file
df = pd.read_csv(csv_file)
# Add a new column 'num' with each value set to the corresponding row index
df['num'] = df.index

# df = df.iloc[0:1000]
# Ensure required columns exist
if "url" not in df.columns or "user_id" not in df.columns:
    raise ValueError("CSV file must contain 'url' and 'user_id' columns.")

def download_image(index, img_url, user_id):
    # Generate the filename using the index and user_id
    filename = f"{index}.jpg"  # Adding an underscore between index and user_id
    img_path = os.path.join(save_folder, filename)

    try:
        response = requests.get(img_url, stream=True, timeout=10)
        response.raise_for_status()

        # Load and save the image
        img = Image.open(BytesIO(response.content))
        img.save(img_path)

        return f"✔ Downloaded: {img_path}"
    except Exception as e:
        return f"❌ Failed: {img_url} - {e}"


# Use multithreading for concurrent downloads
MAX_THREADS = min(32, os.cpu_count() * 2)  # Use twice the number of CPU cores

with ThreadPoolExecutor(max_workers=MAX_THREADS) as executor:
    future_to_url = {
        executor.submit(download_image, index, row["url"], row["user_id"]): row["url"]
        for index, row in df.iterrows()
    }

    for future in as_completed(future_to_url):
        print(future.result())
```

```python
import numpy as np
from PIL import Image, ImageDraw
# … make sure pipe_1 is already defined/imported …

def inpaint_image_1(prompt: str, image_path: str):
    # Load input
    input_image = Image.open(image_path).convert("RGB")
    width, height = input_image.size

    # Desired canvas
    canvas_size = (1024, 1024)

    # 1) Resize original to fit while keeping aspect
    scale = min(canvas_size[0] / width, canvas_size[1] / height)
    new_w, new_h = int(width * scale), int(height * scale)
    resized = input_image.resize((new_w, new_h), Image.LANCZOS)

    # 2) Compute padding amounts
    pad_x = (canvas_size[0] - new_w) // 2
    pad_y = (canvas_size[1] - new_h) // 2
    pad_left, pad_right = pad_x, canvas_size[0] - new_w - pad_x
    pad_top, pad_bottom = pad_y, canvas_size[1] - new_h - pad_y

    # 3) Mirror-pad via NumPy
    arr = np.array(resized)
    padded_arr = np.pad(
        arr,
        ((pad_top, pad_bottom), (pad_left, pad_right), (0, 0)),
        mode='reflect'
    )
    background = Image.fromarray(padded_arr)

    # 4) Create mask: preserve center, inpaint outer
    mask = Image.new("L", canvas_size, 255)
    draw = ImageDraw.Draw(mask)
    draw.rectangle(
        [pad_left, pad_top, pad_left + new_w, pad_top + new_h],
        fill=0
    )

    # 5) Run inpainting pipeline
    result = pipe_1(
        prompt=prompt,
        num_inference_steps=12,
        height=canvas_size[1],
        width=canvas_size[0],
        image=background,
        mask_image=mask
    ).images[0]

    return result

def inpaint_image_2(prompt: str, image_path: str):
    # Load input
    input_image = Image.open(image_path).convert("RGB")
    width, height = input_image.size

    # Desired canvas
    canvas_size = (1024, 1024)

    # 1) Resize original to fit while keeping aspect
    scale = min(canvas_size[0] / width, canvas_size[1] / height)
    new_w, new_h = int(width * scale), int(height * scale)
    resized = input_image.resize((new_w, new_h), Image.LANCZOS)

    # 2) Compute padding amounts
    pad_x = (canvas_size[0] - new_w) // 2
    pad_y = (canvas_size[1] - new_h) // 2
    pad_left, pad_right = pad_x, canvas_size[0] - new_w - pad_x
    pad_top, pad_bottom = pad_y, canvas_size[1] - new_h - pad_y

    # 3) Mirror-pad via NumPy
    arr = np.array(resized)
    padded_arr = np.pad(
        arr,
        ((pad_top, pad_bottom), (pad_left, pad_right), (0, 0)),
        mode='reflect'
    )
    background = Image.fromarray(padded_arr)

    # 4) Create mask: preserve center, inpaint outer
    mask = Image.new("L", canvas_size, 255)
    draw = ImageDraw.Draw(mask)
    draw.rectangle(
        [pad_left, pad_top, pad_left + new_w, pad_top + new_h],
        fill=0
    )

    # 5) Run inpainting pipeline
    result = pipe_2(
        prompt=prompt,
        num_inference_steps=12,
        height=canvas_size[1],
        width=canvas_size[0],
        image=background,
        mask_image=mask
    ).images[0]

    return result
```
